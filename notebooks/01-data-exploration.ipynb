```python
import pandas as pd
import numpy as np

players = pd.read_csv("../data/players.csv")
valuations = pd.read_csv("../data/player_valuations.csv")
appearances = pd.read_csv("../data/appearances.csv")
transfers = pd.read_csv("../data/transfers.csv")
games = pd.read_csv("../data/games.csv")

print("Players:", players.shape)
print("Valuations:", valuations.shape)
print("Appearances:", appearances.shape)
print("Transfers:", transfers.shape)
print("Games:", games.shape)
```

```python
appearances["date"] = pd.to_datetime(appearances["date"])

appearances_with_season = appearances.merge(
    games[["game_id", "season"]],
    on="game_id",
    how="left"
)

player_season_stats = (
    appearances_with_season
    .groupby(["player_id", "season"])
    .agg(
        appearances=("game_id", "count"),
        minutes_played=("minutes_played", "sum"),
        goals=("goals", "sum"),
        assists=("assists", "sum"),
        yellow_cards=("yellow_cards", "sum"),
        red_cards=("red_cards", "sum")
    )
    .reset_index()
)

print(player_season_stats.shape)
print(player_season_stats.head())
```

```python
player_info = players[
    [
        "player_id",
        "date_of_birth",
        "position",
        "sub_position",
        "foot",
        "height_in_cm",
        "country_of_citizenship"
    ]
].copy()

player_season = player_season_stats.merge(
    player_info,
    on="player_id",
    how="left"
)

player_season["date_of_birth"] = pd.to_datetime(
    player_season["date_of_birth"]
)

player_season["season_start"] = pd.to_datetime(
    player_season["season"].astype(str) + "-07-01"
)

player_season["age"] = (
    (player_season["season_start"] - player_season["date_of_birth"])
    .dt.days / 365.25
).round(1)

# Fill missing categorical values
categorical_columns = [
    "position",
    "sub_position",
    "foot",
    "country_of_citizenship"
]

for column in categorical_columns:
    player_season[column] = player_season[column].fillna("Unknown")

print(player_season.shape)
print(player_season[[
    "player_id",
    "season",
    "position",
    "age"
]].head())
```

```python
valuations["date"] = pd.to_datetime(valuations["date"])

# Assign each valuation to a football season.
# July is the start of a new season.
valuations["valuation_season"] = valuations["date"].dt.year

valuations.loc[
    valuations["date"].dt.month < 7,
    "valuation_season"
] -= 1

# Keep the latest valuation for each player in each season
season_valuations = (
    valuations
    .sort_values("date")
    .drop_duplicates(
        ["player_id", "valuation_season"],
        keep="last"
    )
)

print("Season valuations:", season_valuations.shape)
```

```python
target_valuations = season_valuations.copy()

# A valuation in season X becomes the target
# for the player's performance in season X-1.
target_valuations["season"] = (
    target_valuations["valuation_season"] - 1
)

target_valuations = target_valuations.rename(
    columns={
        "market_value_in_eur": "target_market_value_eur"
    }
)

target_valuations = target_valuations[
    [
        "player_id",
        "season",
        "target_market_value_eur"
    ]
]

player_season = player_season.merge(
    target_valuations,
    on=["player_id", "season"],
    how="left"
)

print("Player-season dataset:", player_season.shape)
print(
    "Players with target:",
    player_season["target_market_value_eur"].notna().sum()
)
print(
    "Missing targets:",
    player_season["target_market_value_eur"].isna().sum()
)
```

```python
previous_valuations = season_valuations[
    ["player_id", "valuation_season", "market_value_in_eur"]
].copy()

# A valuation from season X becomes the
# previous value for performance in season X+1.
previous_valuations["season"] = (
    previous_valuations["valuation_season"] + 1
)

previous_valuations = previous_valuations.rename(
    columns={
        "market_value_in_eur": "previous_market_value_eur"
    }
)

previous_valuations = previous_valuations[
    [
        "player_id",
        "season",
        "previous_market_value_eur"
    ]
]

player_season = player_season.merge(
    previous_valuations,
    on=["player_id", "season"],
    how="left"
)

print(
    "Missing previous values:",
    player_season["previous_market_value_eur"].isna().sum()
)
```

```python
previous_growth = season_valuations[
    ["player_id", "valuation_season", "market_value_in_eur"]
].copy()

previous_growth = previous_growth.sort_values(
    ["player_id", "valuation_season"]
)

# Get the valuation before the previous valuation
previous_growth["value_before_previous"] = (
    previous_growth
    .groupby("player_id")["market_value_in_eur"]
    .shift(1)
)

# Calculate historical growth
previous_growth["previous_value_growth"] = (
    (
        previous_growth["market_value_in_eur"]
        - previous_growth["value_before_previous"]
    )
    / previous_growth["value_before_previous"]
)

# Move it forward one season so it represents
# information available before our target season.
previous_growth["season"] = (
    previous_growth["valuation_season"] + 1
)

previous_growth = previous_growth[
    [
        "player_id",
        "season",
        "previous_value_growth"
    ]
]

player_season = player_season.merge(
    previous_growth,
    on=["player_id", "season"],
    how="left"
)

print(
    "Missing previous growth:",
    player_season["previous_value_growth"].isna().sum()
)
```

```python
player_season["goals_per_90"] = (
    player_season["goals"]
    / player_season["minutes_played"]
    * 90
)

player_season["assists_per_90"] = (
    player_season["assists"]
    / player_season["minutes_played"]
    * 90
)

print(
    player_season[
        [
            "goals",
            "assists",
            "minutes_played",
            "goals_per_90",
            "assists_per_90"
        ]
    ].head()
)
```

```python
player_season = player_season.sort_values(
    ["player_id", "season"]
)

player_season["previous_season_goals"] = (
    player_season
    .groupby("player_id")["goals"]
    .shift(1)
)

player_season["previous_season_assists"] = (
    player_season
    .groupby("player_id")["assists"]
    .shift(1)
)

player_season["previous_season_minutes"] = (
    player_season
    .groupby("player_id")["minutes_played"]
    .shift(1)
)

print(
    player_season[
        [
            "player_id",
            "season",
            "goals",
            "previous_season_goals",
            "assists",
            "previous_season_assists",
            "minutes_played",
            "previous_season_minutes"
        ]
    ].head(10)
)
```

```python
player_season = player_season.sort_values(
    ["player_id", "season"]
)

player_season["career_seasons"] = (
    player_season
    .groupby("player_id")
    .cumcount() + 1
)

print(
    player_season[
        ["player_id", "season", "career_seasons"]
    ].head(15)
)
```

```python
import pandas as pd
import numpy as np

players = pd.read_csv("../data/players.csv")
valuations = pd.read_csv("../data/player_valuations.csv")
appearances = pd.read_csv("../data/appearances.csv")
games = pd.read_csv("../data/games.csv")

print("Datasets loaded successfully.")
print("Players:", players.shape)
print("Valuations:", valuations.shape)
print("Appearances:", appearances.shape)
print("Games:", games.shape)
```

```python
appearances["date"] = pd.to_datetime(appearances["date"])

appearances_with_season = appearances.merge(
    games[["game_id", "season"]],
    on="game_id",
    how="left"
)

player_season_stats = (
    appearances_with_season
    .groupby(["player_id", "season"])
    .agg(
        appearances=("game_id", "count"),
        minutes_played=("minutes_played", "sum"),
        goals=("goals", "sum"),
        assists=("assists", "sum"),
        yellow_cards=("yellow_cards", "sum"),
        red_cards=("red_cards", "sum")
    )
    .reset_index()
)

print("Player-season dataset created.")
print("Shape:", player_season_stats.shape)
```

```python
player_info = players[
    [
        "player_id",
        "date_of_birth",
        "position",
        "sub_position",
        "foot",
        "height_in_cm",
        "country_of_citizenship"
    ]
].copy()

player_season = player_season_stats.merge(
    player_info,
    on="player_id",
    how="left"
)

# Convert date of birth
player_season["date_of_birth"] = pd.to_datetime(
    player_season["date_of_birth"]
)

# Use July 1 as the beginning of each football season
player_season["season_start"] = pd.to_datetime(
    player_season["season"].astype(str) + "-07-01"
)

# Calculate player's age at the start of the season
player_season["age"] = (
    player_season["season_start"]
    - player_season["date_of_birth"]
).dt.days / 365.25

player_season["age"] = player_season["age"].round(1)

# Fill missing categorical values
categorical_columns = [
    "position",
    "sub_position",
    "foot",
    "country_of_citizenship"
]

for column in categorical_columns:
    player_season[column] = player_season[column].fillna("Unknown")

print("Player information added.")
print("Shape:", player_season.shape)
```

```python
# Convert valuation dates
valuations["date"] = pd.to_datetime(valuations["date"])

# Assign each valuation to a football season
valuations["valuation_season"] = valuations["date"].dt.year

# January–June belongs to the previous season
valuations.loc[
    valuations["date"].dt.month < 7,
    "valuation_season"
] -= 1

# Keep the latest valuation for each player in each season
season_valuations = (
    valuations
    .sort_values("date")
    .drop_duplicates(
        ["player_id", "valuation_season"],
        keep="last"
    )
)

# Shift the valuation forward by one season.
# Example:
# 2023 valuation → target for the 2022 season
target_valuations = season_valuations.copy()

target_valuations["season"] = (
    target_valuations["valuation_season"] - 1
)

target_valuations = target_valuations.rename(
    columns={
        "market_value_in_eur": "target_market_value_eur"
    }
)

target_valuations = target_valuations[
    [
        "player_id",
        "season",
        "target_market_value_eur"
    ]
]

# Add target to player-season dataset
player_season = player_season.merge(
    target_valuations,
    on=["player_id", "season"],
    how="left"
)

print("Target market value added.")
print("Shape:", player_season.shape)
print(
    "Players with target:",
    player_season["target_market_value_eur"].notna().sum()
)
```

```python
print(
    "Players with target:",
    player_season["target_market_value_eur"].notna().sum()
)
```

```python
previous_valuations = season_valuations[
    [
        "player_id",
        "valuation_season",
        "market_value_in_eur"
    ]
].copy()

# Move the valuation one season forward
previous_valuations["season"] = (
    previous_valuations["valuation_season"] + 1
)

previous_valuations = previous_valuations.rename(
    columns={
        "market_value_in_eur": "previous_market_value_eur"
    }
)

previous_valuations = previous_valuations[
    [
        "player_id",
        "season",
        "previous_market_value_eur"
    ]
]

# Merge into our player-season dataset
player_season = player_season.merge(
    previous_valuations,
    on=["player_id", "season"],
    how="left"
)

print("Previous market value added.")
print(
    "Missing previous values:",
    player_season["previous_market_value_eur"].isna().sum()
)
```

```python
previous_growth = season_valuations[
    [
        "player_id",
        "valuation_season",
        "market_value_in_eur"
    ]
].copy()

# Make sure valuations are in chronological order
previous_growth = previous_growth.sort_values(
    ["player_id", "valuation_season"]
)

# Get the valuation before the previous valuation
previous_growth["value_before_previous"] = (
    previous_growth
    .groupby("player_id")["market_value_in_eur"]
    .shift(1)
)

# Calculate growth
previous_growth["previous_value_growth"] = (
    (
        previous_growth["market_value_in_eur"]
        - previous_growth["value_before_previous"]
    )
    / previous_growth["value_before_previous"]
)

# Align with the season being predicted
previous_growth["season"] = (
    previous_growth["valuation_season"] + 1
)

previous_growth = previous_growth[
    [
        "player_id",
        "season",
        "previous_value_growth"
    ]
]

# Add to player-season dataset
player_season = player_season.merge(
    previous_growth,
    on=["player_id", "season"],
    how="left"
)

print("Previous value growth added.")
print(
    "Missing growth values:",
    player_season["previous_value_growth"].isna().sum()
)
```

```python
player_season["goals_per_90"] = (
    player_season["goals"]
    / player_season["minutes_played"]
    * 90
)

player_season["assists_per_90"] = (
    player_season["assists"]
    / player_season["minutes_played"]
    * 90
)

print("Per-90 statistics added.")
print(
    "Minutes = 0:",
    (player_season["minutes_played"] == 0).sum()
)
```

```python
player_season = player_season.sort_values(
    ["player_id", "season"]
)

player_season["previous_season_goals"] = (
    player_season
    .groupby("player_id")["goals"]
    .shift(1)
)

player_season["previous_season_assists"] = (
    player_season
    .groupby("player_id")["assists"]
    .shift(1)
)

player_season["previous_season_minutes"] = (
    player_season
    .groupby("player_id")["minutes_played"]
    .shift(1)
)

print("Previous-season performance added.")
print(
    "Missing previous-season minutes:",
    player_season["previous_season_minutes"].isna().sum()
)
```
